# Solución de 1:

Sea el mapeo logístico

$$
f_r(x)=rx(1-x),
\qquad x\in[0,1],\quad r\in(0,4],
$$

y sea \(X\) una variable aleatoria cuya distribución es invariante bajo \(f_r\).

Es decir, se cumple

$$
P[X\leq x]=P[f_r(X)\leq x],
\qquad x\in\mathbb{R}.
$$

---

## a) Determinación de $(P[X\leq x])$

Queremos demostrar que

$$
P[X\leq x]
=
P\left[
X\leq
\frac{1}{2}
\left(
1-\sqrt{1-\frac{4x}{r}}
\right)
\right]
+
P\left[
X\geq
\frac{1}{2}
\left(
1+\sqrt{1-\frac{4x}{r}}
\right)
\right].
$$

### Paso 1. Utilizar la invariancia de la distribución

Por la hipótesis de invariancia,

$$
P[X\leq x]
=
P[f_r(X)\leq x].
$$

Como

$$
f_r(X)=rX(1-X),
$$

tenemos

$$
P[X\leq x]
=
P[rX(1-X)\leq x].
$$

---

### Paso 2. Resolver la ecuación $(f_r(X)=x)$

Consideremos primero la ecuación

$$
rX(1-X)=x.
$$

Desarrollando,

$$
rX-rX^2=x.
$$

Pasando todo al mismo lado,

$$
rX^2-rX+x=0.
$$

Esta es una ecuación cuadrática en \(X\). Aplicando la fórmula general,

$$
X=
\frac{r\pm\sqrt{(-r)^2-4rx}}{2r}.
$$

Por lo tanto,

$$
X=
\frac{r\pm\sqrt{r^2-4rx}}{2r}.
$$

Factorizando \(r^2\) dentro de la raíz,

$$
X=
\frac{r\pm r\sqrt{1-\frac{4x}{r}}}{2r}.
$$

Como \(r>0\),

$$
X=
\frac{1}{2}
\left(
1\pm\sqrt{1-\frac{4x}{r}}
\right).
$$

Definimos las dos raíces como

$$
x_1=
\frac{1}{2}
\left(
1-\sqrt{1-\frac{4x}{r}}
\right)
$$

y

$$
x_2=
\frac{1}{2}
\left(
1+\sqrt{1-\frac{4x}{r}}
\right).
$$

Claramente,

$$
x_1\leq x_2.
$$

---

### Paso 3. Determinar el conjunto $(f_r(X)\leq x)$

Tenemos que resolver

$$
rX(1-X)\leq x.
$$

Equivalentemente,

$$
rX^2-rX+x\geq0.
$$

Como \(r>0\), la parábola

$$
rX^2-rX+x
$$

abre hacia arriba.

Por lo tanto, la expresión es no negativa **fuera del intervalo determinado por las dos raíces**. Así,

$$
rX^2-rX+x\geq0
$$

si y solamente si

$$
X\leq x_1
$$

o

$$
X\geq x_2.
$$

Por consiguiente,

$$
\{f_r(X)\leq x\}
=
\{X\leq x_1\}\cup\{X\geq x_2\}.
$$

Como estos dos eventos son disjuntos,

$$
P[f_r(X)\leq x]
=
P[X\leq x_1]
+
P[X\geq x_2].
$$

Sustituyendo las expresiones de \(x_1\) y \(x_2\),

$$
P[f_r(X)\leq x]
=
P\left[
X\leq
\frac12
\left(
1-\sqrt{1-\frac{4x}{r}}
\right)
\right]
+
P\left[
X\geq
\frac12
\left(
1+\sqrt{1-\frac{4x}{r}}
\right)
\right].
$$

Finalmente, utilizando nuevamente la invariancia,

$$
\boxed{
P[X\leq x]
=
P\left[
X\leq
\frac12
\left(
1-\sqrt{1-\frac{4x}{r}}
\right)
\right]
+
P\left[
X\geq
\frac12
\left(
1+\sqrt{1-\frac{4x}{r}}
\right)
\right]
}
$$

que es lo que se quería demostrar.

---

# b) Ecuación funcional para la densidad \(g\)

Del inciso anterior tenemos

$$
P[X\leq x]
=
P[X\leq x_1]+P[X\geq x_2],
$$

donde

$$
x_1=
\frac12
\left(
1-\sqrt{1-\frac{4x}{r}}
\right)
$$

y

$$
x_2=
\frac12
\left(
1+\sqrt{1-\frac{4x}{r}}
\right).
$$

Sea \(G(x)\) la función de distribución de \(X\), es decir,

$$
G(x)=P[X\leq x].
$$

Entonces,

$$
P[X\geq x_2]
=
1-G(x_2)
$$

si \(X\) es una variable continua.

Por lo tanto,

$$
G(x)
=
G(x_1)+1-G(x_2).
$$

Es decir,

$$
G(x)
=
G\left(
\frac12
\left(
1-\sqrt{1-\frac{4x}{r}}
\right)
\right)
+
1-
G\left(
\frac12
\left(
1+\sqrt{1-\frac{4x}{r}}
\right)
\right).
$$

Ahora derivamos ambos lados respecto de \(x\).

Como

$$
g(x)=G'(x),
$$

obtenemos

$$
g(x)
=
g(x_1)\frac{dx_1}{dx}
-
g(x_2)\frac{dx_2}{dx}.
$$

Calculemos cada derivada.

Para \(x_1\),

$$
x_1=
\frac12
\left(
1-\sqrt{1-\frac{4x}{r}}
\right).
$$

Entonces,

$$
\frac{dx_1}{dx}
=
-\frac12
\frac{d}{dx}
\left(
1-\frac{4x}{r}
\right)^{1/2}.
$$

Aplicando la regla de la cadena,

$$
\frac{dx_1}{dx}
=
-\frac12
\left[
\frac12
\left(
1-\frac{4x}{r}
\right)^{-1/2}
\left(-\frac4r\right)
\right].
$$

Por tanto,

$$
\frac{dx_1}{dx}
=
\frac{1}{r\sqrt{1-\frac{4x}{r}}}.
$$

Ahora para \(x_2\),

$$
x_2=
\frac12
\left(
1+\sqrt{1-\frac{4x}{r}}
\right).
$$

Así,

$$
\frac{dx_2}{dx}
=
\frac12
\frac{d}{dx}
\left(
1-\frac{4x}{r}
\right)^{1/2},
$$

y nuevamente por la regla de la cadena,

$$
\frac{dx_2}{dx}
=
\frac12
\left[
\frac12
\left(
1-\frac{4x}{r}
\right)^{-1/2}
\left(-\frac4r\right)
\right].
$$

Por lo tanto,

$$
\frac{dx_2}{dx}
=
-\frac{1}{r\sqrt{1-\frac{4x}{r}}}.
$$

Sustituyendo ambas derivadas,

$$
g(x)
=
g(x_1)
\frac{1}{r\sqrt{1-\frac{4x}{r}}}
-
g(x_2)
\left(
-\frac{1}{r\sqrt{1-\frac{4x}{r}}}
\right).
$$

Entonces,

$$
g(x)
=
\frac{g(x_1)+g(x_2)}
{r\sqrt{1-\frac{4x}{r}}}.
$$

Finalmente, sustituyendo \(x_1\) y \(x_2\),

$$
\boxed{
g(x)
=
\frac{1}
{r\sqrt{1-\frac{4x}{r}}}
\left[
g\left(
\frac12
\left(
1-\sqrt{1-\frac{4x}{r}}
\right)
\right)
+
g\left(
\frac12
\left(
1+\sqrt{1-\frac{4x}{r}}
\right)
\right)
\right]
}
$$

que es la ecuación funcional solicitada.

---

# c) Caso particular \(r=4\)

Partimos de la ecuación obtenida en el inciso anterior:

$$
g(x)
=
\frac{1}
{r\sqrt{1-\frac{4x}{r}}}
\left[
g\left(
\frac12
\left(
1-\sqrt{1-\frac{4x}{r}}
\right)
\right)
+
g\left(
\frac12
\left(
1+\sqrt{1-\frac{4x}{r}}
\right)
\right)
\right].
$$

Para \(r=4\),

$$
1-\frac{4x}{r}
=
1-\frac{4x}{4}

$$

Además,

$$
r=4.
$$

Por lo tanto,

$$
\boxed{
g(x)
=
\frac{1}{4\sqrt{1-x}}
\left[
g\left(
\frac{1-\sqrt{1-x}}{2}
\right)
+

}
$$

para \(0<x<1\).

Esta es la ecuación funcional correspondiente al mapeo logístico

$$
f_4(x)=4x(1-x).
$$

# d) Verificación para la distribución \(\operatorname{Beta}(1/2,1/2)\)

La densidad de una distribución Beta con parámetros \(a\) y \(b\) es

$$
g(x)
=
\frac{1}{B(a,b)}
x^{a-1}(1-x)^{b-1},
\qquad 0<x<1.
$$

Para

$$
a=b=\frac12,
$$

tenemos

$$
g(x)
=
\frac{1}{B\left(\frac12,\frac12\right)}
x^{-1/2}(1-x)^{-1/2}.
$$

Sabemos que

$$
B\left(\frac12,\frac12\right)=\pi.
$$

Por lo tanto,

$$
\boxed{
g(x)=\frac{1}{\pi\sqrt{x(1-x)}}
},
\qquad 0<x<1.
$$

Queremos comprobar que esta función satisface

$$
g(x)
=
\frac{1}{4\sqrt{1-x}}
\left[
g\left(
\frac{1-\sqrt{1-x}}{2}
\right)
+
g\left(
\frac{1+\sqrt{1-x}}{2}
\right)
\right].
$$

Para simplificar la demostración, definamos

$$
s=\sqrt{1-x}.
$$

Entonces,

$$
s^2=1-x
$$

y, por lo tanto,

$$
x=1-s^2.
$$

Definimos

$$
x_1=\frac{1-s}{2},
\qquad
x_2=\frac{1+s}{2}.
$$

---

### Evaluación de \(g(x_1)\)

Tenemos

$$
g(x_1)
=
\frac{1}{\pi\sqrt{x_1(1-x_1)}}.
$$

Como

$$
x_1=\frac{1-s}{2},
$$

entonces

$$
1-x_1
=
1-\frac{1-s}{2}
=
\frac{1+s}{2}.
$$

Por lo tanto,

$$
x_1(1-x_1)
=
\frac{1-s}{2}
\frac{1+s}{2}.
$$

Multiplicando,

$$
x_1(1-x_1)
=
\frac{1-s^2}{4}.
$$

Pero

$$
1-s^2=x,
$$

así que

$$
x_1(1-x_1)=\frac{x}{4}.
$$

Por consiguiente,

$$
g(x_1)
=
\frac{1}
{\pi\sqrt{x/4}}
=
\frac{2}{\pi\sqrt{x}}.
$$

---

### Evaluación de \(g(x_2)\)

De manera análoga,

$$
x_2=\frac{1+s}{2}
$$

y

$$
1-x_2=\frac{1-s}{2}.
$$

Entonces,

$$
x_2(1-x_2)
=
\frac{1+s}{2}
\frac{1-s}{2}
=
\frac{1-s^2}{4}
=
\frac{x}{4}.
$$

Por tanto,

$$
g(x_2)
=
\frac{1}{\pi\sqrt{x/4}}
=
\frac{2}{\pi\sqrt{x}}.
$$

Así,

$$
g(x_1)+g(x_2)
=
\frac{4}{\pi\sqrt{x}}.
$$

---

### Sustitución en la ecuación funcional

El lado derecho de la ecuación funcional es

$$
\frac{1}{4\sqrt{1-x}}
\left[
g(x_1)+g(x_2)
\right].
$$

Sustituyendo,

$$
\frac{1}{4\sqrt{1-x}}
\frac{4}{\pi\sqrt{x}}.
$$

Simplificando,

$$
\frac{1}{\pi\sqrt{x}\sqrt{1-x}}.
$$

Por lo tanto,

$$
\frac{1}{\pi\sqrt{x(1-x)}}.
$$

Pero esta expresión es precisamente \(g(x)\). Es decir,

$$
\frac{1}{4\sqrt{1-x}}
\left[
g(x_1)+g(x_2)
\right]
=
g(x).
$$

Por consiguiente,

$$
\boxed{
g(x)=\frac{1}{\pi\sqrt{x(1-x)}}
}
$$

satisface la ecuación funcional para \(r=4\).

Por lo tanto,

$$
\boxed{
X\sim\operatorname{Beta}\left(\frac12,\frac12\right)
}
$$

es una distribución invariante para el mapeo logístico

$$
f_4(x)=4x(1-x).
$$

# e) Simulación del mapeo logístico y verificación del teorema ergódico

Consideramos

$$
f_4(x)=4x(1-x).
$$

Partiendo de un valor inicial \(x_0\in[0,1]\), generamos la sucesión

$$
x_k=f_4^k(x_0).
$$

Es decir,

$$
x_1=f_4(x_0),
$$

$$
x_2=f_4(x_1),
$$

y en general,

$$
x_k=f_4(x_{k-1}).
$$

Elegimos, por ejemplo,

$$
x_0=0.123456789.
$$

Este valor evita algunos puntos iniciales especiales que pueden producir órbitas periódicas muy cortas.

---

## 1. Teorema ergódico

El teorema ergódico establece que, para una función integrable \(h\),

$$
\lim_{n\to\infty}
\frac1n
\sum_{k=1}^{n}
h(f_4^k(x_0))
=
E[h(X)].
$$

Vamos a comprobarlo para

$$
h(x)=x
$$

y

$$
h(x)=x^2.
$$

---

## 2. Esperanza teórica para \(h(x)=x\)

Tenemos

$$
X\sim\operatorname{Beta}
\left(
\frac12,\frac12
\right).
$$

Para una variable \(X\sim\operatorname{Beta}(a,b)\),

$$
E[X]=\frac{a}{a+b}.
$$

Tomando

$$
a=b=\frac12,
$$

obtenemos

$$
E[X]
=
\frac{\frac12}
{\frac12+\frac12}
=
\frac12.
$$

Por lo tanto,

$$
\boxed{
E[X]=\frac12
}
$$

y el promedio ergódico debe satisfacer

$$
\frac1n\sum_{k=1}^{n}f_4^k(x_0)
\longrightarrow
\frac12.
$$

---

## 3. Esperanza teórica para \(h(x)=x^2\)

Para una distribución Beta,

$$
E[X^2]
=
\frac{a(a+1)}
{(a+b)(a+b+1)}.
$$

Sustituyendo

$$
a=b=\frac12,
$$

tenemos

$$
E[X^2]
=
\frac{
\frac12\left(\frac12+1\right)
}{
\left(\frac12+\frac12\right)
\left(\frac12+\frac12+1\right)
}.
$$

Simplificando,

$$
E[X^2]
=
\frac{
\frac12\cdot\frac32
}{
1\cdot2
}.
$$

Por lo tanto,

$$
E[X^2]
=
\frac{3/4}{2}
=
\frac38.
$$

Así,

$$
\boxed{
E[X^2]=\frac38=0.375
}
$$

y el promedio ergódico debe satisfacer

$$
\frac1n
\sum_{k=1}^{n}
\left(f_4^k(x_0)\right)^2
\longrightarrow
\frac38.
$$

---



```python


'''
El siguiente código implementa directamente el mapeo logístico.
'''

# ---------------------------------------------------------
# Parámetros de la simulación
# ---------------------------------------------------------

x0 = 0.123456789
n = 1_000_000

# ---------------------------------------------------------
# Definición del mapeo logístico para r = 4
# ---------------------------------------------------------

def f4(x):
    return 4*x*(1-x)

# ---------------------------------------------------------
# Generación de la trayectoria
# ---------------------------------------------------------

x = np.empty(n)

# Primer valor de la trayectoria
x_actual = x0

for k in range(n):
    # Aplicamos una vez el mapeo
    x_actual = f4(x_actual)

    # Guardamos el resultado
    x[k] = x_actual

# ---------------------------------------------------------
# Promedios ergódicos
# ---------------------------------------------------------

promedio_x = np.mean(x)
promedio_x2 = np.mean(x**2)

# Valores teóricos
esperanza_x = 1/2
esperanza_x2 = 3/8

# ---------------------------------------------------------
# Resultados
# ---------------------------------------------------------

print("Resultados de la simulación")
print("-----------------------------------")

print(f"Promedio ergódico para h(x) = x:   {promedio_x:.10f}")
print(f"Valor teórico E[X]:                 {esperanza_x:.10f}")
print(f"Error absoluto:                    {abs(promedio_x - esperanza_x):.10f}")

print()

print(f"Promedio ergódico para h(x) = x²:  {promedio_x2:.10f}")
print(f"Valor teórico E[X²]:                {esperanza_x2:.10f}")
print(f"Error absoluto:                    {abs(promedio_x2 - esperanza_x2):.10f}")
```

Al ejecutar el código, los resultados deben ser muy cercanos a

$$
\frac12=0.5
$$

para \(h(x)=x\), y a

$$
\frac38=0.375
$$

para \(h(x)=x^2\).

El resultado exacto dependerá del valor inicial y del número de iteraciones.

Para comprobar visualmente el teorema ergódico, podemos calcular los promedios acumulados

$$
A_n^{(1)}
=
\frac1n
\sum_{k=1}^{n}x_k
$$

y

$$
A_n^{(2)}
=
\frac1n
\sum_{k=1}^{n}x_k^2.
$$

Si el teorema ergódico se cumple, entonces

$$
A_n^{(1)}\longrightarrow\frac12
$$

y

$$
A_n^{(2)}\longrightarrow\frac38.
$$



```python


## 5. Mostrar la convergencia de los promedios

# ---------------------------------------------------------
# Promedios acumulados
# ---------------------------------------------------------

n_valores = np.arange(1, n + 1)

promedios_x = np.cumsum(x) / n_valores
promedios_x2 = np.cumsum(x**2) / n_valores

# ---------------------------------------------------------
# Gráfica para h(x) = x
# ---------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(n_valores, promedios_x, label='Promedio acumulado')

plt.axhline(
    y=esperanza_x,
    linestyle='--',
    label='Valor teórico = 1/2'
)

plt.xlabel('n')
plt.ylabel('Promedio')
plt.title('Convergencia del promedio ergódico para h(x) = x')
plt.legend()
plt.grid()

plt.show()

# ---------------------------------------------------------
# Gráfica para h(x) = x²
# ---------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(n_valores, promedios_x2, label='Promedio acumulado')

plt.axhline(
    y=esperanza_x2,
    linestyle='--',
    label='Valor teórico = 3/8'
)

plt.xlabel('n')
plt.ylabel('Promedio')
plt.title('Convergencia del promedio ergódico para h(x) = x²')
plt.legend()
plt.grid()

plt.show()

```

Estas gráficas permiten observar que, conforme aumenta \(n\), los promedios acumulados se aproximan a sus respectivos valores teóricos.

---
Comparación final:

Los valores teóricos son

$$
E[X]=\frac12=0.5
$$

y

$$
E[X^2]=\frac38=0.375.
$$

Por otra parte, mediante la simulación obtenemos

$$
\frac1n\sum_{k=1}^{n}f_4^k(x_0)
\approx 0.5
$$

y

$$
\frac1n\sum_{k=1}^{n}
\left(f_4^k(x_0)\right)^2
\approx 0.375.
$$

Por lo tanto,

$$
\boxed{
\frac1n\sum_{k=1}^{n}f_4^k(x_0)
\longrightarrow
\frac12
}
$$

y

$$
\boxed{
\frac1n\sum_{k=1}^{n}
\left(f_4^k(x_0)\right)^2
\longrightarrow
\frac38
}.
$$

Esto proporciona evidencia numérica de que el teorema ergódico se cumple para el mapeo logístico \(f_4\) y la distribución invariante

$$
\boxed{
X\sim\operatorname{Beta}
\left(
\frac12,\frac12
\right).
}
$$

---
Conclusión:

Para el mapeo logístico

$$
f_r(x)=rx(1-x),
$$

la invariancia de la distribución permite transformar la condición

$$
P[X\leq x]
=
P[f_r(X)\leq x]
$$

en una expresión que involucra las dos raíces de la ecuación

$$
rx(1-x)=x.
$$

A partir de esta expresión se obtiene la ecuación funcional para la densidad invariante,

$$
g(x)
=
\frac{1}
{r\sqrt{1-\frac{4x}{r}}}
\left[
g\left(
\frac12
\left(
1-\sqrt{1-\frac{4x}{r}}
\right)
\right)
+
g\left(
\frac12
\left(
1+\sqrt{1-\frac{4x}{r}}
\right)
\right)
\right].
$$

Para \(r=4\), se verifica que

$$
\boxed{
g(x)=\frac{1}{\pi\sqrt{x(1-x)}}
}
$$

satisface dicha ecuación. Esta es precisamente la densidad de

$$
\operatorname{Beta}
\left(
\frac12,\frac12
\right).
$$

Finalmente, mediante simulación numérica se verifica que los promedios ergódicos convergen a

$$
E[X]=\frac12
$$

y

$$
E[X^2]=\frac38,
$$

confirmando numéricamente el resultado teórico.

Muestre que las variables aleatorias

U\sim\operatorname{Uniforme}(0,1)

# Solución de 2:

Muestre que las variables aleatorias

$$
U\sim\operatorname{Uniforme}(0,1)
$$

y

$$
X\sim\operatorname{Beta}(\gamma,\gamma),
\qquad \gamma>0,
$$

son invariantes bajo el mapeo

$$
f(x)=1-x.
$$

Recordemos que una distribución es invariante bajo \(f\) si

$$
P[X\leq x]=P[f(X)\leq x],
\qquad x\in\mathbb R.
$$

Demostraremos la invariancia para cada una de las dos distribuciones.

---

# 1. Caso \(U\sim\operatorname{Uniforme}(0,1)\)

Sea

$$
U\sim\operatorname{Uniforme}(0,1)
$$

y consideremos

$$
f(U)=1-U.
$$

Queremos demostrar que

$$
P[U\leq x]=P[1-U\leq x].
$$

## Paso 1. Función de distribución de \(U\)

La función de distribución de una variable uniforme en \((0,1)\) es

$$
F_U(x)=P[U\leq x]
=
\begin{cases}
0, & x\leq 0,\\[4pt]
x, & 0<x<1,\\[4pt]
1, & x\geq1.
\end{cases}
$$

Ahora calcularemos la función de distribución de \(1-U\).

---

## Paso 2. Calcular \(P[1-U\leq x]\)

Tenemos

$$
1-U\leq x.
$$

Restando \(1\) en ambos lados,

$$
-U\leq x-1.
$$

Al multiplicar por \(-1\), debemos invertir la desigualdad:

$$
U\geq1-x.
$$

Por lo tanto,

$$
P[1-U\leq x]
=
P[U\geq1-x].
$$

Como \(U\) es una variable continua,

$$
P[U\geq1-x]
=
1-P[U<1-x].
$$

Equivalentemente, para \(0<x<1\),

$$
P[U\geq1-x]
=
1-(1-x)
=
x.
$$

Por lo tanto,

$$
P[1-U\leq x]=x,
\qquad 0<x<1.
$$

Pero también

$$
P[U\leq x]=x,
\qquad 0<x<1.
$$

Así,

$$
P[U\leq x]
=
P[1-U\leq x].
$$

Para completar la demostración, revisamos los casos extremos.

### Si \(x\leq0\)

Como \(U\in(0,1)\),

$$
P[U\leq x]=0.
$$

Además,

$$
1-U\in(0,1),
$$

por lo que

$$
P[1-U\leq x]=0.
$$

### Si \(x\geq1\)

Tenemos

$$
P[U\leq x]=1
$$

y también

$$
P[1-U\leq x]=1.
$$

Por lo tanto, para todo \(x\in\mathbb R\),

$$
\boxed{
P[U\leq x]=P[1-U\leq x]
}.
$$

Así,

$$
\boxed{
U\sim\operatorname{Uniforme}(0,1)
}
$$

es invariante bajo el mapeo

$$
\boxed{f(x)=1-x}.
$$

---

# 2. Caso \(X\sim\operatorname{Beta}(\gamma,\gamma)\)

Ahora consideremos

$$
X\sim\operatorname{Beta}(\gamma,\gamma),
\qquad \gamma>0.
$$

Queremos demostrar que

$$
P[X\leq x]
=
P[1-X\leq x].
$$

En este caso resulta particularmente útil observar la simetría de la distribución Beta cuando ambos parámetros son iguales.

---

## Paso 1. Escribir la función de densidad

La densidad de una distribución Beta \((\gamma,\gamma)\) es

$$
g(x)
=
\frac{1}{B(\gamma,\gamma)}
x^{\gamma-1}(1-x)^{\gamma-1},
\qquad 0<x<1,
$$

donde \(B(\gamma,\gamma)\) es la función Beta.

Recordemos que

$$
B(a,b)
=
\int_0^1
t^{a-1}(1-t)^{b-1}\,dt.
$$

En nuestro caso,

$$
B(\gamma,\gamma)
=
\int_0^1
t^{\gamma-1}(1-t)^{\gamma-1}\,dt.
$$

---

## Paso 2. Observar la simetría de la densidad

Evaluemos la densidad en \(1-x\):

$$
g(1-x)
=
\frac{1}{B(\gamma,\gamma)}
(1-x)^{\gamma-1}
\left[1-(1-x)\right]^{\gamma-1}.
$$

Simplificando,

$$
1-(1-x)=x.
$$

Por lo tanto,

$$
g(1-x)
=
\frac{1}{B(\gamma,\gamma)}
(1-x)^{\gamma-1}
x^{\gamma-1}.
$$

Como la multiplicación es conmutativa,

$$
g(1-x)
=
\frac{1}{B(\gamma,\gamma)}
x^{\gamma-1}
(1-x)^{\gamma-1}.
$$

Por consiguiente,

$$
\boxed{
g(1-x)=g(x)
}.
$$

Es decir, la densidad de una distribución

$$
\operatorname{Beta}(\gamma,\gamma)
$$

es simétrica alrededor de

$$
x=\frac12.
$$

---

## Paso 3. Calcular la distribución de \(1-X\)

Sea

$$
Y=1-X.
$$

Queremos demostrar que \(Y\) tiene la misma distribución que \(X\).

Para \(0<y<1\), la función de distribución de \(Y\) es

$$
F_Y(y)
=
P[Y\leq y].
$$

Como

$$
Y=1-X,
$$

tenemos

$$
F_Y(y)
=
P[1-X\leq y].
$$

Despejando \(X\),

$$
1-X\leq y
$$

implica

$$
-X\leq y-1.
$$

Multiplicando por \(-1\) e invirtiendo la desigualdad,

$$
X\geq1-y.
$$

Por lo tanto,

$$
F_Y(y)
=
P[X\geq1-y].
$$

Como \(X\) es una variable continua,

$$
P[X\geq1-y]
=
1-F_X(1-y).
$$

Entonces,

$$
F_Y(y)
=
1-F_X(1-y).
$$

Ahora escribimos la función de distribución de \(X\) mediante su densidad:

$$
F_X(1-y)
=
\int_0^{1-y}
g(t)\,dt.
$$

Por lo tanto,

$$
F_Y(y)
=
1-
\int_0^{1-y}
g(t)\,dt.
$$

Como la densidad integra uno,

$$
1=
\int_0^1g(t)\,dt.
$$

Así,

$$
F_Y(y)
=
\int_{1-y}^1g(t)\,dt.
$$

Hacemos el cambio de variable

$$
u=1-t.
$$

Entonces,

$$
du=-dt.
$$

Cuando

$$
t=1-y,
$$

tenemos

$$
u=y,
$$

y cuando

$$
t=1,
$$

tenemos

$$
u=0.
$$

Por lo tanto,

$$
F_Y(y)
=
\int_y^0g(1-u)(-du).
$$

Es decir,

$$
F_Y(y)
=
\int_0^y g(1-u)\,du.
$$

Pero demostramos anteriormente que

$$
g(1-u)=g(u).
$$

Por consiguiente,

$$
F_Y(y)
=
\int_0^y g(u)\,du.
$$

Por definición,

$$
\int_0^y g(u)\,du=F_X(y).
$$

Entonces,

$$
\boxed{
F_Y(y)=F_X(y)
}.
$$

Como

$$
Y=1-X,
$$

obtenemos

$$
\boxed{
1-X\overset{d}{=}X
}.
$$

Es decir, \(X\) y \(1-X\) tienen exactamente la misma distribución.

---

# 3. Demostración directamente mediante la invariancia

También podemos expresar la demostración utilizando directamente la definición solicitada.

Para

$$
X\sim\operatorname{Beta}(\gamma,\gamma),
$$

tenemos

$$
P[f(X)\leq x]
=
P[1-X\leq x].
$$

Como

$$
1-X\leq x
\iff
X\geq1-x,
$$

se obtiene

$$
P[f(X)\leq x]
=
P[X\geq1-x].
$$

Por la simetría de la distribución,

$$
P[X\geq1-x]
=
P[X\leq x].
$$

Por lo tanto,

$$
\boxed{
P[f(X)\leq x]=P[X\leq x]
}.
$$

Así,

$$
\boxed{
X\sim\operatorname{Beta}(\gamma,\gamma)
}
$$

es invariante bajo

$$
\boxed{
f(x)=1-x
}.
$$

---

# 4. Conclusión

Hemos demostrado que las dos distribuciones son invariantes bajo el mapeo

$$
f(x)=1-x.
$$

En el caso de la uniforme,

$$
\boxed{
U\sim\operatorname{Uniforme}(0,1)
\quad\Longrightarrow\quad
1-U\sim\operatorname{Uniforme}(0,1)
}.
$$

En el caso de la distribución Beta,

$$
\boxed{
X\sim\operatorname{Beta}(\gamma,\gamma)
\quad\Longrightarrow\quad
1-X\sim\operatorname{Beta}(\gamma,\gamma)
},
\qquad \gamma>0.
$$

La razón fundamental en el segundo caso es que su densidad

$$
g(x)
=
\frac{1}{B(\gamma,\gamma)}
x^{\gamma-1}(1-x)^{\gamma-1}
$$

es simétrica:

$$
\boxed{
g(1-x)=g(x)
}.
$$

Por lo tanto, el mapeo

$$
\boxed{f(x)=1-x}
$$

refleja la distribución respecto al punto \(x=1/2\), pero no modifica su distribución de probabilidad.

# Solución de 3:

Consideremos el mapeo ``carpa'' definido por

f_{\text{carpa}}(x)

2x, & x\leq \dfrac12,\\[6pt]
2(1-x), & x>\dfrac12.

Este mapeo transforma el intervalo \([0,1]\) nuevamente en \([0,1]\).

---

# a) Invariancia de la distribución Uniforme\((0,1)\)

Sea

$$
X\sim\operatorname{Uniforme}(0,1).
$$

Queremos demostrar que la distribución de \(X\) es invariante bajo el mapeo carpa. Es decir, debemos verificar que

$$
P[X\leq x]
=
P[f_{\text{carpa}}(X)\leq x],
\qquad x\in\mathbb R.
$$

La función de distribución de \(X\) es

$$
F_X(x)
=
P[X\leq x]
=
\begin{cases}
0, & x\leq0,\\[4pt]
x, & 0<x<1,\\[4pt]
1, & x\geq1.
\end{cases}
$$

Por lo tanto, basta demostrar la igualdad para

$$
0<x<1.
$$

---

## Paso 1. Dividir el evento según la definición del mapeo

Para \(0<x<1\),

$$
P[f_{\text{carpa}}(X)\leq x]
$$

puede separarse dependiendo de si \(X\leq1/2\) o \(X>1/2\):

$$
\begin{aligned}
P[f_{\text{carpa}}(X)\leq x]
={}&
P[f_{\text{carpa}}(X)\leq x,\ X\leq\tfrac12]
\\
&+
P[f_{\text{carpa}}(X)\leq x,\ X>\tfrac12].
\end{aligned}
$$

En el primer intervalo,

$$
f_{\text{carpa}}(X)=2X,
$$

mientras que en el segundo,

$$
f_{\text{carpa}}(X)=2(1-X).
$$

Por lo tanto,

$$
\begin{aligned}
P[f_{\text{carpa}}(X)\leq x]
={}&
P[2X\leq x,\ X\leq\tfrac12]
\\
&+
P[2(1-X)\leq x,\ X>\tfrac12].
\end{aligned}
$$

---

## Paso 2. Primer término

Consideremos

$$
P[2X\leq x,\ X\leq\tfrac12].
$$

La desigualdad

$$
2X\leq x
$$

equivale a

$$
X\leq\frac{x}{2}.
$$

Como \(0<x<1\), tenemos

$$
0<\frac{x}{2}<\frac12.
$$

Por lo tanto, la condición \(X\leq1/2\) queda automáticamente satisfecha cuando

$$
X\leq\frac{x}{2}.
$$

Así,

$$
P[2X\leq x,\ X\leq\tfrac12]
=
P\left[X\leq\frac{x}{2}\right].
$$

Como \(X\) es uniforme,

$$
P\left[X\leq\frac{x}{2}\right]
=
\frac{x}{2}.
$$

Por lo tanto,

$$
\boxed{
P[2X\leq x,\ X\leq\tfrac12]
=
\frac{x}{2}
}.
$$

---

## Paso 3. Segundo término

Ahora consideremos

$$
P[2(1-X)\leq x,\ X>\tfrac12].
$$

Tenemos

$$
2(1-X)\leq x.
$$

Dividiendo entre \(2\),

$$
1-X\leq\frac{x}{2}.
$$

Entonces,

$$
-X\leq\frac{x}{2}-1.
$$

Multiplicando por \(-1\) e invirtiendo la desigualdad,

$$
X\geq1-\frac{x}{2}.
$$

Como \(0<x<1\),

$$
\frac12<1-\frac{x}{2}<1.
$$

Por lo tanto, la condición \(X>1/2\) se satisface automáticamente cuando

$$
X\geq1-\frac{x}{2}.
$$

Así,

$$
P[2(1-X)\leq x,\ X>\tfrac12]
=
P\left[X\geq1-\frac{x}{2}\right].
$$

Como \(X\sim\operatorname{Uniforme}(0,1)\),

$$
P\left[X\geq1-\frac{x}{2}\right]
=
1-\left(1-\frac{x}{2}\right)
=
\frac{x}{2}.
$$

Por lo tanto,

$$
\boxed{
P[2(1-X)\leq x,\ X>\tfrac12]
=
\frac{x}{2}
}.
$$

---

## Paso 4. Sumar ambos términos

Tenemos

$$
\begin{aligned}
P[f_{\text{carpa}}(X)\leq x]
&=
\frac{x}{2}+\frac{x}{2}\\
&=x.
\end{aligned}
$$

Pero, como

$$
X\sim\operatorname{Uniforme}(0,1),
$$

para \(0<x<1\),

$$
P[X\leq x]=x.
$$

Por lo tanto,

$$
P[f_{\text{carpa}}(X)\leq x]
=
P[X\leq x].
$$

Para \(x\leq0\), ambos lados son \(0\), y para \(x\geq1\), ambos lados son \(1\).

Así, para todo \(x\in\mathbb R\),

$$
\boxed{
P[f_{\text{carpa}}(X)\leq x]
=
P[X\leq x]
}.
$$

Por consiguiente,

$$
\boxed{
X\sim\operatorname{Uniforme}(0,1)
}
$$

es invariante bajo el mapeo carpa.

# b) Simulación del mapeo carpa

Ahora programaremos el mapeo

$$
f_{\text{carpa}}(x)
=
\begin{cases}
2x, & x\leq\dfrac12,\\[6pt]
2(1-x), & x>\dfrac12.
\end{cases}
$$

Elegiremos inicialmente

$$
x_0=0.123456789
$$

y calcularemos sucesivamente

$$
x_{k+1}=f_{\text{carpa}}(x_k).
$$

Es decir,

$$
x_1=f_{\text{carpa}}(x_0),
$$

$$
x_2=f_{\text{carpa}}(x_1),
$$

y en general,

$$
x_k=f_{\text{carpa}}^k(x_0).
$$

---



```python
## Código en Python

import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Definición del mapeo carpa
# ---------------------------------------------------------

def f_carpa(x):
    if x <= 0.5:
        return 2*x
    else:
        return 2*(1-x)


# ---------------------------------------------------------
# Parámetros
# ---------------------------------------------------------

x0 = 0.123456789
n = 100

# Arreglo para almacenar la trayectoria
trayectoria = np.empty(n + 1)

# Condición inicial
trayectoria[0] = x0

# ---------------------------------------------------------
# Iteración del mapeo
# ---------------------------------------------------------

for k in range(n):
    trayectoria[k + 1] = f_carpa(trayectoria[k])


# ---------------------------------------------------------
# Mostrar algunos valores
# ---------------------------------------------------------

print("Primeros valores de la trayectoria:")
print("-----------------------------------")

for k in range(20):
    print(f"x_{k} = {trayectoria[k]:.10f}")


## Visualización de la trayectoria

#Podemos representar \(x_k\) en función de \(k\):

# ---------------------------------------------------------
# Gráfica de la trayectoria
# ---------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    range(n + 1),
    trayectoria,
    marker='o',
    markersize=3,
    linewidth=1
)

plt.xlabel('Iteración k')
plt.ylabel(r'$x_k$')
plt.title('Trayectoria del mapeo carpa')
plt.grid()

plt.show()

```

La gráfica muestra cómo evoluciona el sistema dinámico discreto

$$
x_{k+1}=f_{\text{carpa}}(x_k).
$$

El mapeo tiene una estructura de ``carpa'' porque su gráfica está formada por dos segmentos de recta:

$$
f_{\text{carpa}}(x)=2x
$$

para

$$
0\leq x\leq\frac12,
$$

y

$$
f_{\text{carpa}}(x)=2(1-x)
$$

para

$$
\frac12<x\leq1.
$$

El punto máximo se alcanza en

$$
x=\frac12,
$$

donde

$$
f_{\text{carpa}}\left(\frac12\right)=1.
$$

# Caso particular: \(x_0=\frac{1}{2^n}\)

Ahora consideremos la condición inicial

$$
x_0=\frac{1}{2^n}.
$$

Este caso es especialmente importante debido a la representación finita de los números en una computadora.

Mientras el valor se encuentre en

$$
0\leq x_k\leq\frac12,
$$

el mapeo carpa está dado por

$$
f_{\text{carpa}}(x_k)=2x_k.
$$

Por lo tanto,

$$
x_{k+1}=2x_k.
$$

Partiendo de

$$
x_0=\frac{1}{2^n},
$$

obtenemos

$$
x_1
=
2\left(\frac{1}{2^n}\right)
=
\frac{1}{2^{n-1}}.
$$

Después,

$$
x_2
=
2x_1
=
2\left(\frac{1}{2^{n-1}}\right)
=
\frac{1}{2^{n-2}}.
$$

Continuando de esta manera,

$$
x_k
=
\frac{1}{2^{n-k}},
$$

siempre que \(k\leq n\).

En particular,

$$
x_{n-1}
=
\frac12.
$$

Entonces,

$$
x_n
=
f_{\text{carpa}}\left(\frac12\right)
=
2\left(\frac12\right)
=
1.
$$

Finalmente,

$$
x_{n+1}
=
f_{\text{carpa}}(1)
=
2(1-1)
=
0.
$$

Y una vez que llegamos a \(0\),

$$
f_{\text{carpa}}(0)=0.
$$

Por lo tanto,

$$
\boxed{
x_0=\frac{1}{2^n}
\quad\Longrightarrow\quad
x_{n+1}=0
}
$$

y todos los valores posteriores permanecen en \(0\):

$$
x_{n+2}=x_{n+3}=\cdots=0.
$$

$$

Podemos observarlo mediante Python:


```python

# ---------------------------------------------------------
# Caso x0 = 1 / 2^n
# ---------------------------------------------------------

n = 10

x0 = 1 / (2**n)

num_iteraciones = n + 5

trayectoria_potencia = np.empty(num_iteraciones + 1)

trayectoria_potencia[0] = x0

for k in range(num_iteraciones):
    trayectoria_potencia[k + 1] = f_carpa(trayectoria_potencia[k])


# Mostrar resultados
print(f"x0 = 1/2^{n}")
print("-----------------------------------")

for k, valor in enumerate(trayectoria_potencia):
    print(f"x_{k} = {valor:.15f}")


#También podemos visualizar la trayectoria:


plt.figure(figsize=(10, 5))

plt.plot(
    range(num_iteraciones + 1),
    trayectoria_potencia,
    marker='o'
)

plt.xlabel('Iteración k')
plt.ylabel(r'$x_k$')
plt.title(r'Trayectoria para $x_0=1/2^{10}$')
plt.grid()

plt.show()


```

# Efecto de la representación finita en computadora

En una computadora, los números reales no se representan con precisión infinita. Se utiliza una cantidad finita de bits para almacenar cada número.

Esto es especialmente importante para el mapeo carpa porque, mientras

$$
x_k\leq\frac12,
$$

tenemos

$$
x_{k+1}=2x_k.
$$

Por ejemplo, si un número se representa aproximadamente como

$$
x_k=0.000101101\ldots,
$$

multiplicarlo por \(2\) equivale, en representación binaria, a desplazar los bits hacia la izquierda.

Por esta razón, el mapeo carpa puede ir eliminando progresivamente los bits menos significativos de la representación finita del número.

En particular, para

$$
x_0=\frac1{2^n},
$$

tenemos en representación binaria

$$
x_0=0.\underbrace{00\cdots0}_{n-1}1_2.
$$

Cada aplicación del mapeo, mientras el valor esté en la primera mitad del intervalo, multiplica por \(2\), desplazando el \(1\) una posición hacia la izquierda:

$$
\frac1{2^n}
\rightarrow
\frac1{2^{n-1}}
\rightarrow
\frac1{2^{n-2}}
\rightarrow\cdots
\rightarrow
\frac12
\rightarrow
1
\rightarrow
0.
$$

Por ello, debido a la representación finita de los números en la computadora, eventualmente la trayectoria puede llegar exactamente a

$$
0,
$$

a partir de la cual queda atrapada en el punto fijo

$$
f_{\text{carpa}}(0)=0.
$$

---

# Conclusión

En el inciso (a), demostramos que si

$$
X\sim\operatorname{Uniforme}(0,1),
$$

entonces

$$
P[f_{\text{carpa}}(X)\leq x]
=
P[X\leq x]
$$

para todo \(x\in\mathbb R\). Por lo tanto,

$$
\boxed{
\operatorname{Uniforme}(0,1)
\text{ es una distribución invariante del mapeo carpa.}
}
$$

En el inciso (b), al iterar

$$
x_{k+1}=f_{\text{carpa}}(x_k),
$$

obtenemos la trayectoria del sistema dinámico discreto.

Para la condición inicial particular

$$
\boxed{
x_0=\frac1{2^n}
},
$$

la trayectoria satisface

$$
x_k=\frac1{2^{n-k}},
\qquad k=0,\ldots,n,
$$

y posteriormente

$$
x_{n+1}=0.
$$

Así,

$$
\boxed{
\frac1{2^n}
\longrightarrow
\frac1{2^{n-1}}
\longrightarrow
\cdots
\longrightarrow
\frac12
\longrightarrow
1
\longrightarrow
0
\longrightarrow
0
\longrightarrow\cdots
}
$$

lo cual ilustra cómo la representación finita de los números en una computadora puede provocar que una trayectoria que matemáticamente se comportaría de manera diferente termine entrando en una órbita periódica o, en este caso particular, quede atrapada en el punto fijo \(0\).

# Solución de 4:

Sea \(X\) una variable aleatoria cuya función cumulativa de distribución es

$$
F_{\text{carpa}}(x)
=
\begin{cases}
0, & x\leq 0,\\[6pt]
\dfrac{x}{4}, & 0<x\leq 2,\\[8pt]
\dfrac{1}{2}, & 2<x\leq 3,\\[8pt]
1-\dfrac{\exp(-x+3)}{2}, & 3<x.
\end{cases}
$$

Queremos demostrar que \(X\) puede simularse mediante

$$
X=
\begin{cases}
X_1, & \text{con probabilidad }\dfrac12,\\[8pt]
X_2, & \text{con probabilidad }\dfrac12,
\end{cases}
$$

donde

$$
X_1\sim\operatorname{Uniforme}(0,2)
$$

y

$$
X_2=3+E,
$$

con

$$
E\sim\operatorname{Exponencial}(1).
$$

La demostración se hará construyendo la función de distribución de la variable aleatoria propuesta y verificando que coincide con \(F_{\text{carpa}}\).

---

# 1. Interpretación de la función de distribución dada

La función de distribución es

$$
F_{\text{carpa}}(x)
=
\begin{cases}
0, & x\leq0,\\
\dfrac{x}{4}, & 0<x\leq2,\\
\dfrac12, & 2<x\leq3,\\
1-\dfrac{\exp(-x+3)}2, & x>3.
\end{cases}
$$

Primero observemos qué sucede en los puntos importantes.

En \(x=2\),

$$
F_{\text{carpa}}(2)
=
\frac24
=
\frac12.
$$

Después, para

$$
2<x\leq3,
$$

la función permanece constante:

$$
F_{\text{carpa}}(x)=\frac12.
$$

Esto significa que no se acumula probabilidad entre \(2\) y \(3\).

Finalmente, cuando \(x>3\),

$$
F_{\text{carpa}}(x)
=
1-\frac{e^{-x+3}}2.
$$

Cuando \(x\to3^+\),

$$
F_{\text{carpa}}(x)
\to
1-\frac{e^0}{2}
=
1-\frac12
=
\frac12.
$$

Por lo tanto, la primera parte de la distribución acumula una probabilidad total de

$$
\boxed{\frac12},
$$

mientras que la parte posterior a \(3\) acumula la probabilidad restante

$$
\boxed{\frac12}.
$$

Esto sugiere una mezcla de dos distribuciones, cada una con peso \(1/2\).

---

# 2. Primera componente: \(X_1\sim\operatorname{Uniforme}(0,2)\)

Sea

$$
X_1\sim\operatorname{Uniforme}(0,2).
$$

Su función de distribución es

$$
F_{X_1}(x)
=
\begin{cases}
0, & x\leq0,\\[4pt]
\dfrac{x}{2}, & 0<x\leq2,\\[6pt]
1, & x>2.
\end{cases}
$$

Como en nuestra construcción \(X_1\) se seleccionará con probabilidad \(1/2\), su contribución a la función de distribución será

$$
\frac12F_{X_1}(x).
$$

Para

$$
0<x\leq2,
$$

tenemos

$$
\frac12F_{X_1}(x)
=
\frac12\left(\frac{x}{2}\right)
=
\frac{x}{4}.
$$

Esto coincide exactamente con el primer tramo no trivial de \(F_{\text{carpa}}\):

$$
\boxed{
F_{\text{carpa}}(x)=\frac{x}{4},
\qquad 0<x\leq2.
}
$$

Además, como \(X_1\) siempre toma valores entre \(0\) y \(2\), esta componente ha acumulado toda su probabilidad \(1/2\) al llegar a \(x=2\).

En efecto,

$$
\frac12F_{X_1}(2)
=
\frac12(1)
=
\frac12.
$$

---

# 3. Segunda componente: \(X_2=3+E\)

Ahora consideremos

$$
E\sim\operatorname{Exponencial}(1).
$$

La función de distribución de una variable exponencial con parámetro \(\lambda=1\) es

$$
F_E(e)
=
\begin{cases}
0, & e\leq0,\\[4pt]
1-e^{-e}, & e>0.
\end{cases}
$$

Definimos

$$
X_2=3+E.
$$

Por lo tanto,

$$
X_2\geq3.
$$

Queremos obtener su función de distribución.

---

## Paso 1. Calcular \(F_{X_2}(x)\)

Por definición,

$$
F_{X_2}(x)
=
P[X_2\leq x].
$$

Como

$$
X_2=3+E,
$$

tenemos

$$
F_{X_2}(x)
=
P[3+E\leq x].
$$

Restando \(3\),

$$
F_{X_2}(x)
=
P[E\leq x-3].
$$

### Si \(x\leq3\)

Entonces

$$
x-3\leq0.
$$

Como \(E\) es una variable exponencial no negativa,

$$
P[E\leq x-3]=0.
$$

Por lo tanto,

$$
F_{X_2}(x)=0,
\qquad x\leq3.
$$

### Si \(x>3\)

Entonces

$$
x-3>0.
$$

Utilizando la función de distribución exponencial,

$$
F_{X_2}(x)
=
1-e^{-(x-3)}.
$$

Como

$$
-(x-3)=-x+3,
$$

obtenemos

$$
\boxed{
F_{X_2}(x)
=
1-e^{-x+3},
\qquad x>3.
}
$$

---

# 4. Mezcla de las dos distribuciones

Consideremos ahora una variable aleatoria \(X\) definida por

$$
X=
\begin{cases}
X_1, & \text{con probabilidad }\dfrac12,\\[6pt]
X_2, & \text{con probabilidad }\dfrac12.
\end{cases}
$$

En otras palabras, primero elegimos una de las dos distribuciones con probabilidades iguales:

$$
P(X=X_1)=\frac12,
$$

$$
P(X=X_2)=\frac12.
$$

La función de distribución de una mezcla es

$$
F_X(x)
=
\frac12F_{X_1}(x)
+
\frac12F_{X_2}(x).
$$

Por lo tanto,

$$
\boxed{
F_X(x)
=
\frac12F_{X_1}(x)
+
\frac12F_{X_2}(x)
}.
$$

Ahora verificaremos cada intervalo.

---

# 5. Verificación por intervalos

## Caso 1: \(x\leq0\)

Tenemos

$$
F_{X_1}(x)=0
$$

y

$$
F_{X_2}(x)=0.
$$

Por lo tanto,

$$
F_X(x)
=
\frac12(0)+\frac12(0)
=
0.
$$

Así,

$$
\boxed{
F_X(x)=0,\qquad x\leq0.
}
$$

Coincide con \(F_{\text{carpa}}\).

---

## Caso 2: \(0<x\leq2\)

Para \(X_1\),

$$
F_{X_1}(x)=\frac{x}{2}.
$$

Para \(X_2\), como \(x\leq2<3\),

$$
F_{X_2}(x)=0.
$$

Entonces,

$$
F_X(x)
=
\frac12\left(\frac{x}{2}\right)
+
\frac12(0).
$$

Por lo tanto,

$$
F_X(x)
=
\frac{x}{4}.
$$

Así,

$$
\boxed{
F_X(x)=\frac{x}{4},
\qquad 0<x\leq2.
}
$$

Nuevamente coincide con \(F_{\text{carpa}}\).

---

## Caso 3: \(2<x\leq3\)

En este intervalo,

$$
F_{X_1}(x)=1,
$$

porque \(X_1\leq2<x\).

Por otro lado,

$$
F_{X_2}(x)=0,
$$

porque \(X_2\geq3\).

Entonces,

$$
F_X(x)
=
\frac12(1)+\frac12(0).
$$

Por lo tanto,

$$
\boxed{
F_X(x)=\frac12,
\qquad 2<x\leq3.
}
$$

Esto coincide con el tercer tramo de \(F_{\text{carpa}}\).

---

## Caso 4: \(x>3\)

En este caso,

$$
F_{X_1}(x)=1,
$$

porque \(X_1\leq2<x\).

Además,

$$
F_{X_2}(x)
=
1-e^{-x+3}.
$$

Por lo tanto,

$$
\begin{aligned}
F_X(x)
&=
\frac12(1)
+
\frac12(1-e^{-x+3})\\
&=
\frac12+\frac12-\frac12e^{-x+3}\\
&=
1-\frac{e^{-x+3}}2.
\end{aligned}
$$

Por consiguiente,

$$
\boxed{
F_X(x)
=
1-\frac{e^{-x+3}}2,
\qquad x>3.
}
$$

Esto coincide exactamente con el último tramo de \(F_{\text{carpa}}\).

---

# 6. Comparación final

Hemos obtenido

$$
F_X(x)
=
\begin{cases}
0, & x\leq0,\\[6pt]
\dfrac{x}{4}, & 0<x\leq2,\\[8pt]
\dfrac12, & 2<x\leq3,\\[8pt]
1-\dfrac{\exp(-x+3)}2, & x>3.
\end{cases}
$$

Esta es exactamente la función de distribución proporcionada:

$$
F_X(x)=F_{\text{carpa}}(x).
$$

Por lo tanto, la variable aleatoria \(X\) puede ser simulada mediante la mezcla

$$
\boxed{
X=
\begin{cases}
X_1, & \text{con probabilidad }\dfrac12,\\[8pt]
X_2, & \text{con probabilidad }\dfrac12,
\end{cases}
}
$$

donde

$$
\boxed{
X_1\sim\operatorname{Uniforme}(0,2)
}
$$

y

$$
\boxed{
X_2=3+E,
\qquad E\sim\operatorname{Exponencial}(1).
}
$$

---

# 7. Interpretación para simulación

El resultado también nos proporciona un procedimiento para generar observaciones de \(X\).

Primero generamos una variable

$$
U\sim\operatorname{Uniforme}(0,1).
$$

Después:

* Si \(U\leq1/2\), generamos \(X_1\sim\operatorname{Uniforme}(0,2)\) y tomamos \(X=X_1\).
* Si \(U>1/2\), generamos \(E\sim\operatorname{Exponencial}(1)\) y tomamos

  $$
  X=3+E.
  $$

En forma de algoritmo:

$$
\boxed{
\begin{array}{ll}
U\leq\frac12
&\Rightarrow X=X_1,\quad X_1\sim\operatorname{Uniforme}(0,2),\\[6pt]
U>\frac12
&\Rightarrow X=3+E,\quad E\sim\operatorname{Exponencial}(1).
\end{array}
}
$$

Por lo tanto, la distribución puede simularse sin necesidad de utilizar directamente la función de distribución por partes.

# Conclusión

La función de distribución dada puede interpretarse como una **mezcla de dos distribuciones con pesos \(1/2\)**.

La primera componente es

$$
\operatorname{Uniforme}(0,2),
$$

que aporta una probabilidad total de \(1/2\).

La segunda componente es una distribución exponencial desplazada,

$$
3+\operatorname{Exponencial}(1),
$$

que aporta la probabilidad restante \(1/2\).

Por lo tanto,

$$
\boxed{
X\overset{d}{=}
\begin{cases}
X_1, & \text{con probabilidad }\frac12,\\[6pt]
3+E, & \text{con probabilidad }\frac12,
\end{cases}
}
$$

con

$$
X_1\sim\operatorname{Uniforme}(0,2),
\qquad
E\sim\operatorname{Exponencial}(1).
$$

Esto demuestra la afirmación del ejercicio.
